## Question 3 (Heavy tails)

The standard deviation $\sigma$ of a spread's daily changes describes its risk completely only if those changes are close to normal. This question checks.

### (a)

Let $X \sim N(\mu, \sigma^2)$, with moment generating function

$$
M(t) = \mathbb{E}\left[e^{tX}\right] = \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right).
$$

Using $\mathbb{E}[X^k] = M^{(k)}(0)$, compute $\mathbb{E}[X]$, $\mathbb{E}[X^2]$, $\mathbb{E}[X^3]$ and $\mathbb{E}[X^4]$. Then show that

$$
\mathbb{E}\left[(X-\mu)^3\right] = 0
\quad\text{and}\quad
\mathbb{E}\left[(X-\mu)^4\right] = 3\sigma^4,
$$

so that the kurtosis $\mathbb{E}\left[(X-\mu)^4\right]/\sigma^4$ equals 3 for every normal distribution.

**Answer.**

We have $X \sim N(\mu, \sigma^2)$, with moment generating function

$$
M(t) = \mathbb{E}\left[e^{tX}\right] = \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right).
$$

We know that

$$ \mathbb{E}[X^k] = M^k(0) $$

so we find

\begin{align*}
\mathbb{E}[X]
& = M'(0) = \left. \frac{d}{dt} M(t) \right|_{t = 0} \\
& = \left. \frac{d}{dt} \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right|_{t = 0} \\
& = \left. (\mu + \sigma^2 t) \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right|_{t = 0} \\
& = (\mu + \sigma^2 0) \exp\left(\mu 0 + \tfrac{1}{2}\sigma^2 0^2\right) \\
& = \mu \\
\mathbb{E}[X^2]
& = M''(0) = \left. \frac{d}{dt} M'(t) \right|_{t = 0} \\
& = \left. \frac{d}{dt} (\mu + \sigma^2 t) \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right|_{t = 0} \\
& = \left. \sigma^2 \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) + (\mu + \sigma^2 t)^2 \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right)\right|_{t = 0} \\
& = \left. \sigma^2 \exp\left(\mu 0 + \tfrac{1}{2}\sigma^2 0^2\right) + (\mu + \sigma^2 0)^2 \exp\left(\mu 0 + \tfrac{1}{2}\sigma^2 0^2\right)\right|_{t = 0} \\
& = \sigma^2 + \mu^2 \\
\mathbb{E}[X^3]
& = M'''(0) = \left. \frac{d}{dt} M''(t) \right|_{t = 0} \\
& = \left. \frac{d}{dt} \left[ \left(\sigma^2 + (\mu + \sigma^2 t)^2\right) \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right] \right|_{t = 0} \\
& = \left. \left[ 2(\mu + \sigma^2 t)\sigma^2 \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) + \left(\sigma^2 + (\mu + \sigma^2 t)^2\right) (\mu + \sigma^2 t)  \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right] \right|_{t = 0} \\
& = 2\mu\sigma^2  + \left(\sigma^2 + \mu^2 \right) \mu \\
& = \mu^3 + 3\mu\sigma^2 \\
\mathbb{E}[X^4]
& = M^{(4)}(0) = \left. \frac{d}{dt} M'''(t) \right|_{t = 0} \\
& = \left. \frac{d}{dt} \left[ \left( 2(\mu + \sigma^2 t)\sigma^2 + \left(\sigma^2 + (\mu + \sigma^2 t)^2\right) (\mu + \sigma^2 t) \right)  \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right] \right|_{t = 0} \\
& = \left. \frac{d}{dt} \left[ \left( 3\sigma^2(\mu + \sigma^2 t) + (\mu + \sigma^2 t)^3 \right)  \exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right] \right|_{t = 0} \\
& = \left. \left[
\left( 3\sigma^4 + 3 (\mu + \sigma^2 t)^2\sigma^2 \right)
\exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right) \right.\right. \\ & \quad\quad\quad + \left. \left.
\left( 3\sigma^2(\mu + \sigma^2 t) + (\mu + \sigma^2 t)^3 \right)
(\mu + \sigma^2 t)\exp\left(\mu t + \tfrac{1}{2}\sigma^2 t^2\right)
\right] \right|_{t = 0} \\
& = \left( 3\sigma^4 + 3 \mu^2\sigma^2 \right) +
\left( 3\mu\sigma^2 + \mu^3 \right)
\mu \\
& = \mu^4 + 6\mu^2\sigma^2 + 3\sigma^4
\end{align*}

\begin{align*}
\mathbb{E}\left[(X-\mu)^3\right]
& = \mathbb{E}\left[ X^3 - 3X^2\mu + 3X\mu^2 - \mu^3 \right] \\
& = \mathbb{E}\left[ X^3 \right] - 3\mu\mathbb{E}\left[ X^2 \right] + 3\mu^2\mathbb{E}\left[ X \right] - \mu^3 \\
& = \mu^3 + 3\mu\sigma^2 - 3\mu(\sigma^2 + \mu^2) + 3\mu^3 - \mu^3 \\
& = \mu^3 + 3\mu\sigma^2 - 3\mu\sigma^2 - 3\mu^3 + 3\mu^3 - \mu^3 \\
& = 0
\end{align*}

\begin{align*}
\mathbb{E}\left[(X-\mu)^4\right]
& = \mathbb{E}\left[X^4 - 4 X^3\mu + 6X^2\mu^2 - 4 X\mu^3 + \mu^4 \right] \\
& = \mathbb{E}\left[ X^4 \right] - 4\mu \mathbb{E}\left[ X^3 \right] + 6\mu^2 \mathbb{E}\left[ X^2 \right] - 4\mu^3 \mathbb{E}\left[ X \right] + \mu^4 \\
& = \mu^4 + 6\mu^2\sigma^2 + 3\sigma^4 - 4\mu (\mu^3 + 3\mu\sigma^2) + 6\mu^2 (\mu^2 + \sigma^2) - 4\mu^3 \mu + \mu^4 \\
& = \mu^4 + 6\mu^2\sigma^2 + 3\sigma^4 - 4\mu^4 - 12\mu^2\sigma^2 + 6\mu^4 + 6\mu^2\sigma^2 - 4\mu^4 + \mu^4 \\
& = 3\sigma^4
\end{align*}


### (b)

For a standardized variable $Z = (X-\mu)/\sigma$ the kurtosis is $\mathbb{E}[Z^4]$. Because $z^4$ grows so fast (a $4\sigma$ day counts $4^4 = 256$ times as much as a $1\sigma$ day), the kurtosis is driven almost entirely by the largest moves. Kurtosis above 3 means more extreme days than a normal distribution with the same variance: heavy tails. Under a normal distribution, $|Z| > 4$ happens on about 0.006% of days, roughly once in 60 years of trading.

Compute the kurtosis of daily changes in `NG[H][J0]`, `CL[F][G0]` and `HE[J][M0]`:

1. Build each spread, and keep instances 2000–2025 and days with `td_to_ltd` from $-250$ to $-2$.
2. Within each instance, compute the daily price changes $S_t - S_{t-1}$ of the spread settlement. Use price differences, not percentage changes: a spread's price can be near zero or negative.
3. Standardize each instance's changes by that instance's own mean and standard deviation, pool all instances, and compute the average of $z^4$.

Using each instance's whole-sample mean and standard deviation is fine here: this describes the data and is not a trading signal. Report the kurtosis of each spread.

**Answer.**

```python
# terminal command: python -m futspread build "NG[H][J0]" "CL[F][G0]" "HE[J][M0]"

'''using liquid months | building 3 combo(s)
  [1/3] NG[H][J0]               29 instances  13,897 rows
  [2/3] CL[F][G0]               28 instances  13,470 rows
  [3/3] HE[J][M0]               28 instances   8,127 rows
3/3 combos had usable instances'''
```

```python
import os
os.environ['FUTSPREAD_ROOT'] = '/Users/robli/Documents/Columbia/2026Fall/DDF/data/futspread'
```

```python
from futspread import load_combo

df = load_combo('NG[H][J0]')
df[['instance', 'date', 'settlement', 'td_to_ltd']].head()
```

```python
def pooled_kurt(combo):
    df = load_combo(combo)
    mask = ((df['instance'] >= 2000) & (df['instance'] <= 2025) & (df['td_to_ltd'] >= -250) & (df['td_to_ltd'] <= -2))
    d = df[mask].copy()
    d = d.sort_values(['instance', 'date'])
    d['price_changes'] = d.groupby('instance')['settlement'].diff()
    d = d.dropna(subset=['price_changes'])
    pooled_changes = d.groupby('instance')['price_changes']
    d['z'] = (d['price_changes'] - pooled_changes.transform('mean')) / pooled_changes.transform('std')
    avg_kurt = (d['z'] ** 4).mean()
    return avg_kurt

combos = ['NG[H][J0]', 'CL[F][G0]', 'HE[J][M0]']
for combo in combos:
    print("Pooled Kurtosis for daily spread changes for " + combo + ": ", round(pooled_kurt(combo), 4))
```

> All three spreads have kurtosis well above the normal value of 3, so their
daily changes are heavy-tailed: extreme days occur far more often than a normal
distribution with the same variance would imply, and $\sigma$ alone understates
their risk.
