```python
import pandas as pd
from pathlib import Path
from futspread import universe

root = Path(os.environ['FUTSPREAD_ROOT'])
u = universe()

rows = []
for code, tic in zip(u['contrcode'], u['ticker']):
    p = root / 'raw' / 'prices' / f'{code}.parquet'
    if p.exists():
        d = pd.read_parquet(p)['date']
        rows.append((tic, d.min(), d.max(), len(d)))
    else:
        rows.append((tic, None, None, 0))

cov = pd.DataFrame(rows, columns=['ticker', 'first', 'last', 'n_bars'])
print(len(u), 'series in universe;', (cov.n_bars > 0).sum(), 'with price files')
cov.sort_values('last')
```

```python
excluded = {'ARE', 'RBB', 'HOB', 'GZ', 'LBR', 'LBS'}
short = cov[~cov.ticker.isin(excluded) & (cov['last'] < '2025-12-31')]
print(short.ticker.tolist())
```
